```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
tips = sns.load_dataset("tips")
```

```python
tips.head()
```

```python
tips.info()
```

```python
tips.describe()
```

```python
tips.isnull().sum()
```

```python
tips.drop_duplicates(inplace=True)
```

```python
tips.shape
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")
plt.title("Total Bill vs Tip by Gender")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="day")
plt.title("Total Bill vs Tip by Day")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time")
plt.title("Total Bill vs Tip by Time")
plt.show()
```

Executive Summary (Tips)

• Higher restaurant bills generally result in higher tips.
• Dinner customers tend to have larger bills and tips than lunch customers.
• Tipping patterns are similar for both genders with only minor differences.

```python
titanic = sns.load_dataset("titanic")
```

```python
titanic.head()
```

```python
titanic.info()
```

```python
titanic.describe()
```

```python
titanic.isnull().sum()
```

```python
titanic["age"] = titanic["age"].fillna(titanic["age"].median())
```

```python
titanic["embarked"] = titanic["embarked"].fillna(titanic["embarked"].mode()[0])
```

```python
titanic.drop(columns="deck", inplace=True)
```

```python
titanic.drop_duplicates(inplace=True)
```

```python
titanic["AgeGroup"] = pd.cut(
    titanic["age"],
    bins=[0,18,35,60,100],
    labels=["Child","Young Adult","Adult","Senior"]
)
```

```python
sns.barplot(data=titanic, x="class", y="survived")
plt.title("Survival Rate by Passenger Class")
plt.show()
```

```python
sns.barplot(data=titanic, x="sex", y="survived")
plt.title("Survival Rate by Gender")
plt.show()
```

```python
sns.barplot(data=titanic, x="AgeGroup", y="survived")
plt.title("Survival Rate by Age Group")
plt.show()
```
